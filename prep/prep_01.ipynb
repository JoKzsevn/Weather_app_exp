```python
import os
from dotenv import load_dotenv
import requests
from pathlib import Path
import json
```

# API connection test

```python
# Load environment variables from .env
load_dotenv()

# Get API key
api_key = os.getenv("XWEATHER_API_KEY")

if not api_key:
    raise ValueError("API key not found!")

print("API key loaded successfully!")
```

# Reading endpoint data

```python
# Load environment variables
load_dotenv()

# Get API credentials
client_id = os.getenv("CLIENT_ID")
client_secret = os.getenv("SECRET_KEY")

if not client_id or not client_secret:
    raise ValueError("Xweather credentials not found!")

# API endpoint
url = "https://data.api.xweather.com/conditions/rotterdam,nl"

# Parameters
params = {
    "client_id": client_id,
    "client_secret": client_secret
}

# Send request
response = requests.get(url, params=params, timeout=10)

# Check for HTTP errors
response.raise_for_status()

# Convert JSON response
data = response.json()

# Display response
print(data)
```

# Saving output as json

```python
# Define folder
folder = Path("../data")

# Create folder if it doesn't exist
folder.mkdir(parents=True, exist_ok=True)

# Define file path
file_path = folder / "Rotterdam_weather_data.json"

# Save JSON
with open(file_path, "w", encoding="utf-8") as f:
    json.dump(data, f, indent=4, ensure_ascii=False)

print(f"JSON saved to: {file_path.resolve()}")
```

```python
import json
from pathlib import Path
import streamlit as st

# Page configuration
st.set_page_config(
    page_title="Rotterdam Weather Dashboard",
    page_icon="🌤️",
    layout="wide"
)

# Load JSON
file_path = Path("../data/Rotterdam_weather_data.json")

with open(file_path, "r", encoding="utf-8") as f:
    data = json.load(f)

# Extract weather data
weather_data = data["response"][0]
period = weather_data["periods"][0]

# Extract values
date_time = period["dateTimeISO"]
temperature = period["tempC"]
feels_like = period["feelslikeC"]
humidity = period["humidity"]
wind_direction = period["windDir"]
visibility = period["visibilityKM"]
weather = period["weather"]
weather_primary = period["weatherPrimary"]
timezone = weather_data.get("profile", {}).get("tz", "Europe/Amsterdam")

# Dashboard
st.title("🌤️ Rotterdam Weather Dashboard")

st.write(f"📅 **Date and Time:** {date_time}")
st.write(f"🌍 **Timezone:** {timezone}")

st.divider()

# First row
col1, col2, col3 = st.columns(3)

col1.metric("🌡️ Temperature", f"{temperature} °C")
col2.metric("🌡️ Feels Like", f"{feels_like} °C")
col3.metric("💧 Humidity", f"{humidity}%")

# Second row
col4, col5 = st.columns(2)

col4.metric("🧭 Wind Direction", wind_direction)
col5.metric("👁️ Visibility", f"{visibility} km")

st.divider()

# Weather conditions
st.subheader("☁️ Weather Conditions")

st.write(f"**Weather:** {weather}")
st.write(f"**Weather Primary:** {weather_primary}")
```
